```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed/ipl_cleaned.csv")
```

```python
team_matches = df["batting_team"].value_counts()

team_matches
```

```python
plt.figure(figsize=(12,6))

team_matches.plot(kind="bar")

plt.title("Matches Played by Teams")
plt.xlabel("Team")
plt.ylabel("Matches")

plt.xticks(rotation=45)

plt.show()
```

```python
team_wins = df["winner"].value_counts()

team_wins
```

```python
plt.figure(figsize=(12,6))

team_wins.plot(kind="bar", color="green")

plt.title("Matches Won by Teams")
plt.xlabel("Team")
plt.ylabel("Wins")

plt.xticks(rotation=45)

plt.show()
```

```python
toss = df["toss_winner"].value_counts()

toss
```

```python
plt.figure(figsize=(12,6))

toss.plot(kind="bar", color="orange")

plt.title("Toss Winners")
plt.xlabel("Team")
plt.ylabel("Toss Wins")

plt.xticks(rotation=45)

plt.show()
```

```python
summary = pd.DataFrame({
    "Matches Played": team_matches,
    "Matches Won": team_wins
}).fillna(0)

summary["Win Percentage"] = (
    summary["Matches Won"] /
    summary["Matches Played"] * 100
).round(2)

summary.sort_values("Win Percentage", ascending=False)
```

```python
summary.to_csv("reports/team_analysis.csv", index=True)

print("Team Analysis Completed")
```
