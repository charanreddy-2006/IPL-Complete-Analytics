```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed/ipl_cleaned.csv")
```

```python
purple_cap = (
    df[df["player_out"].notna()]
    .groupby("bowler")["player_out"]
    .count()
    .sort_values(ascending=False)
)

purple_cap.head(10)
```

```python
top10 = purple_cap.head(10)

plt.figure(figsize=(12,6))

top10.plot(kind="bar", color="purple")

plt.title("Top 10 Wicket Takers (Purple Cap)")
plt.xlabel("Bowler")
plt.ylabel("Wickets")

plt.xticks(rotation=45)

plt.show()
```

```python
import os

os.makedirs("../reports", exist_ok=True)

purple_cap.reset_index().to_csv(
    "../reports/purple_cap.csv",
    index=False
)

print("Purple Cap analysis completed.")
```
