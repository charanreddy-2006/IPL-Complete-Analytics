```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed/ipl_cleaned.csv")
```

```python
orange_cap = (
    df.groupby("batter")["runs_scored"]
      .sum()
      .sort_values(ascending=False)
)

orange_cap.head(10)
```

```python
top10 = orange_cap.head(10)

plt.figure(figsize=(12,6))

top10.plot(kind="bar")

plt.title("Top 10 Run Scorers (Orange Cap)")
plt.xlabel("Batter")
plt.ylabel("Total Runs")

plt.xticks(rotation=45)

plt.show()
```

```python
orange_cap.reset_index().head(10)
```

```python
import os

# Create reports folder if it doesn't exist
os.makedirs("../reports", exist_ok=True)

# Save the file
orange_cap.reset_index().to_csv(
    os.path.join("../reports", "orange_cap.csv"),
    index=False
)

print("Orange Cap analysis completed.")
```
